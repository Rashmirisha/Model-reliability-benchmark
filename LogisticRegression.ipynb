```python
data = data.dropna(subset=["runtime"])
```

```python
x = data[["budget", "revenue", "runtime", "vote_average", "vote_count"]]

y = (data["popularity"] >= data["popularity"].median()).astype(int)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
print(y_pred[:10])
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test,y_pred)
print(accuracy)
```

```python
x_test_perturbed = x_test.copy().astype(float)

x_test_perturbed.iloc[:, 0] = x_test.iloc[:, 0] * 1.10
y_pred_perturbed = model.predict(x_test_perturbed)
```

```python
accuracy_perturbed = accuracy_score(y_test, y_pred_perturbed)

print(accuracy_perturbed)
```
